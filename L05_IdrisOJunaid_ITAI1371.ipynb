## Part 2: Handling Missing Values (Imputation)**Concept:** Most machine learning models cannot handle missing values (`NaN`). We must deal with them. Dropping the rows is an option, but you lose data. A better way is **imputation**, which means filling in the missing values with a calculated guess.Common imputation strategies:*   **Mean:** Fill with the average value. Good for normally distributed data.*   **Median:** Fill with the middle value. Better for skewed data or data with outliers (like `Fare`).*   **Mode:** Fill with the most frequent value. Used for categorical data.

### Task 1: Impute the `Age` Column

The `Age` column is missing many values. Since age can be skewed by very young or very old passengers, using the **median** is a robust choice.

**Your Task:** Calculate the median of the `Age` column and use the `.fillna()` method to replace the missing values.

```python
# 1. Calculate the median of the Age column.
median_age = df["Age"].median()

# 2. Fill the missing values in Age with the median age.
# This keeps the dataset size the same instead of dropping rows with missing age values.
df["Age"] = df["Age"].fillna(median_age)

# 3. Verify that there are no more missing values in Age.
print("Median age used for imputation:", median_age)
print("Missing values in Age after imputation:")
print(df["Age"].isnull().sum())
```

## Part 3: Encoding Categorical Features**Concept:** Machine learning models are mathematical, so they need numbers, not text. We need to convert categorical columns (like 'Sex' and 'Embarked') into a numerical format. The most common method is **One-Hot Encoding**.One-Hot Encoding takes a column with `N` categories and turns it into `N` new columns, each with a `1` or `0`. For example, the 'Sex' column (`male`, `female`) becomes two new columns: `Sex_male` and `Sex_female`.Pandas has a convenient function called `pd.get_dummies()` that does this for us.

### Task 2: One Hot Encode Categorical Columns

**Your Task:** Use `pd.get_dummies()` to encode the `Sex` and `Embarked` columns. The original text columns are replaced by numeric dummy columns.

```python
# Use one hot encoding to convert the Sex and Embarked categorical columns into numeric columns.
# drop_first=True removes one category from each encoded feature to reduce redundant information.
encoded_df = pd.get_dummies(df, columns=["Sex", "Embarked"], drop_first=True)

# Display the first few rows of the encoded DataFrame to confirm the new columns were created.
print(encoded_df.head())
print("\nEncoded columns:")
print(encoded_df.columns.tolist())
```

## Part 4: Feature Scaling**Concept:** Many models are sensitive to the scale of the features. For example, `Age` (from 0-80) and `Fare` (from 0-512) are on very different scales. This can cause the model to incorrectly believe that `Fare` is a more important feature simply because its values are larger.**Feature Scaling** solves this by putting all features on a similar scale. A common method is **Standardization** (`StandardScaler` in scikit-learn), which rescales the data to have a mean of 0 and a standard deviation of 1.**Important:** You only scale your numerical features, not your target variable or your newly encoded categorical columns.

### Task 3: Scale the `Age` and `Fare` Columns

**Your Task:** Use `StandardScaler` from `sklearn.preprocessing` to scale the `Age` and `Fare` columns.

```python
from sklearn.preprocessing import StandardScaler

# 1. Create an instance of StandardScaler.
scaler = StandardScaler()

# 2. Select only the numerical feature columns that need scaling.
columns_to_scale = ["Age", "Fare"]

# 3. Fit the scaler to the selected columns and transform those columns.
encoded_df[columns_to_scale] = scaler.fit_transform(encoded_df[columns_to_scale])

# 4. Display the first few rows to confirm that Age and Fare were standardized.
print(encoded_df.head())
print("\nScaled feature summary:")
print(encoded_df[columns_to_scale].describe())
```

**Instructions:** Answer the following questions in this markdown cell.

1. **Why is it often better to impute missing values with the median instead of the mean?**

   It is often better to use the median when the data is skewed or contains outliers. The mean can be pulled too high or too low by extreme values, while the median represents the middle value and is more stable. For example, in the Titanic dataset, a few very old passengers could make the average age less representative of the typical passenger age. Using the median gives a more reliable replacement value for missing ages.

2. **Explain in your own words what One Hot Encoding does and why it is necessary.**

   One Hot Encoding changes text based categories into numeric columns that a machine learning model can understand. For example, the `Sex` column contains values such as male and female, and `Embarked` contains port categories. A model cannot directly calculate with those words, so One Hot Encoding creates new columns with values such as 0 or 1 to show whether each row belongs to a category. This keeps the category information while making the data usable for machine learning.

3. **Would you need to apply Feature Scaling to a Decision Tree model? Why or why not?**

   Feature scaling is usually not required for a Decision Tree model. A Decision Tree makes splits by comparing values within a feature, such as checking whether `Age` is less than a certain number. Because the tree is based on split points and not distance calculations, large values in one feature do not automatically dominate smaller scale features. Feature scaling is more important for models that use distances or gradient based optimization, such as K Nearest Neighbors, Logistic Regression, or Support Vector Machines.
